```python
def get_kaggle_video_path(relative_path):

    path = relative_path.replace("\\", "/")

    if "data/raw/cel/" in path:

        relative = path.split(
            "data/raw/cel/",
            1
        )[1]

        return os.path.join(
            CELEB_ROOT,
            relative
        )

    elif "data/raw/FaceForensics++_C23/" in path:

        relative = path.split(
            "data/raw/FaceForensics++_C23/",
            1
        )[1]

        return os.path.join(
            FF_ROOT,
            relative
        )

    return None

print("✅ Path conversion ready")
```

```python
import cv2
import pandas as pd
import numpy as np
from PIL import Image
from tqdm.auto import tqdm
import os

# -----------------------------
# Settings for TEST
# -----------------------------

NUM_TEST_VIDEOS = 10

# Process every 10th frame
FRAME_INTERVAL = 10

# Minimum face confidence
MIN_CONFIDENCE = 0.90

results = []

# Take first 10 videos
test_df = df.head(NUM_TEST_VIDEOS)

print("Videos to test:", len(test_df))
print("=" * 60)


# -----------------------------
# Process videos
# -----------------------------

for video_index, row in tqdm(
    test_df.iterrows(),
    total=len(test_df),
    desc="Step 4 test"
):

    csv_path = row["video_path"]

    # Convert CSV path → Kaggle dataset path
    video_path = get_kaggle_video_path(csv_path)

    if video_path is None or not os.path.exists(video_path):
        print("❌ Missing:", csv_path)
        continue

    # Open video
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("❌ Could not open:", video_path)
        continue

    frame_number = 0
    frames_checked = 0
    faces_found = 0

    # -----------------------------
    # Read frames
    # -----------------------------

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        # Only process every 10th frame
        if frame_number % FRAME_INTERVAL != 0:
            frame_number += 1
            continue

        frames_checked += 1

        # BGR → RGB
        rgb_frame = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        # Convert to PIL image
        image = Image.fromarray(rgb_frame)

        # -----------------------------
        # Detect faces using MTCNN
        # -----------------------------

        boxes, probabilities = mtcnn.detect(image)

        if boxes is None or probabilities is None:
            frame_number += 1
            continue

        # -----------------------------
        # Keep confident faces
        # -----------------------------

        valid_faces = []

        for box, confidence in zip(
            boxes,
            probabilities
        ):

            if confidence is None:
                continue

            if confidence < MIN_CONFIDENCE:
                continue

            x1, y1, x2, y2 = box

            width = max(0, x2 - x1)
            height = max(0, y2 - y1)

            area = width * height

            valid_faces.append(
                (area, box, confidence)
            )

        if not valid_faces:
            frame_number += 1
            continue

        # -----------------------------
        # Select largest face
        # -----------------------------

        valid_faces.sort(
            key=lambda x: x[0],
            reverse=True
        )

        area, box, confidence = valid_faces[0]

        x1, y1, x2, y2 = box

        # Frame dimensions
        h, w = rgb_frame.shape[:2]

        # Make coordinates valid
        x1 = max(0, int(x1))
        y1 = max(0, int(y1))
        x2 = min(w, int(x2))
        y2 = min(h, int(y2))

        if x2 <= x1 or y2 <= y1:
            frame_number += 1
            continue

        # -----------------------------
        # Crop face
        # -----------------------------

        face = rgb_frame[y1:y2, x1:x2]

        if face.size == 0:
            frame_number += 1
            continue

        # -----------------------------
        # Create output folder
        # -----------------------------

        video_name = os.path.splitext(
            os.path.basename(csv_path)
        )[0]

        video_output_dir = os.path.join(
            FACES_DIR,
            video_name
        )

        os.makedirs(
            video_output_dir,
            exist_ok=True
        )

        # -----------------------------
        # Save face image
        # -----------------------------

        face_filename = (
            f"frame_{frame_number:06d}.jpg"
        )

        face_path = os.path.join(
            video_output_dir,
            face_filename
        )

        Image.fromarray(face).save(
            face_path,
            quality=95
        )

        faces_found += 1

        # -----------------------------
        # Save metadata
        # -----------------------------

        results.append({
            "video_path": csv_path,
            "frame_number": frame_number,
            "face_path": face_path,
            "confidence": float(confidence),
            "x1": x1,
            "y1": y1,
            "x2": x2,
            "y2": y2,
            "face_width": x2 - x1,
            "face_height": y2 - y1
        })

        frame_number += 1

    cap.release()

    print(
        f"{video_index + 1}/{len(test_df)} | "
        f"frames checked: {frames_checked} | "
        f"faces saved: {faces_found}"
    )


# -----------------------------
# Save metadata
# -----------------------------

results_df = pd.DataFrame(results)

metadata_path = os.path.join(
    OUTPUT_DIR,
    "step4_test_metadata.csv"
)

results_df.to_csv(
    metadata_path,
    index=False
)


# -----------------------------
# Final report
# -----------------------------

print("\n" + "=" * 60)
print("✅ STEP 4 TEST COMPLETE")
print("=" * 60)

print("Videos tested:", len(test_df))
print("Face crops saved:", len(results_df))
print("Metadata file:", metadata_path)

if len(results_df) > 0:

    print(
        "Average confidence:",
        round(
            results_df["confidence"].mean(),
            4
        )
    )

    print("\nFirst 5 results:")
    display(results_df.head())
```

```python
import cv2
import os
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

FRAME_INTERVAL = 10
MIN_CONFIDENCE = 0.90

results = []

# IMPORTANT: use the new balanced dataset
test_df = test_videos.copy()

print("Videos to process:", len(test_df))
print("=" * 60)

for video_index, row in tqdm(
    test_df.iterrows(),
    total=len(test_df),
    desc="Step 5 face extraction"
):

    csv_path = row["video_path"]
    video_path = get_kaggle_video_path(csv_path)

    label = int(row["label"])
    label_name = row["label_name"]

    if video_path is None or not os.path.exists(video_path):
        print("❌ Missing:", csv_path)
        continue

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("❌ Could not open:", video_path)
        continue

    frame_number = 0
    frames_checked = 0
    faces_found = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        if frame_number % FRAME_INTERVAL != 0:
            frame_number += 1
            continue

        frames_checked += 1

        rgb_frame = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        image = Image.fromarray(rgb_frame)

        boxes, probabilities = mtcnn.detect(image)

        if boxes is None or probabilities is None:
            frame_number += 1
            continue

        valid_faces = []

        for box, confidence in zip(
            boxes,
            probabilities
        ):

            if confidence is None:
                continue

            if confidence < MIN_CONFIDENCE:
                continue

            x1, y1, x2, y2 = box

            width = max(0, x2 - x1)
            height = max(0, y2 - y1)

            area = width * height

            valid_faces.append(
                (area, box, confidence)
            )

        if not valid_faces:
            frame_number += 1
            continue

        # Largest face
        valid_faces.sort(
            key=lambda x: x[0],
            reverse=True
        )

        area, box, confidence = valid_faces[0]

        x1, y1, x2, y2 = box

        h, w = rgb_frame.shape[:2]

        x1 = max(0, int(x1))
        y1 = max(0, int(y1))
        x2 = min(w, int(x2))
        y2 = min(h, int(y2))

        if x2 <= x1 or y2 <= y1:
            frame_number += 1
            continue

        face = rgb_frame[y1:y2, x1:x2]

        if face.size == 0:
            frame_number += 1
            continue

        # Create a unique folder using label + video name
        video_name = os.path.splitext(
            os.path.basename(csv_path)
        )[0]

        video_output_dir = os.path.join(
            FACES_DIR,
            label_name,
            video_name
        )

        os.makedirs(
            video_output_dir,
            exist_ok=True
        )

        face_filename = (
            f"frame_{frame_number:06d}.jpg"
        )

        face_path = os.path.join(
            video_output_dir,
            face_filename
        )

        Image.fromarray(face).save(
            face_path,
            quality=95
        )

        faces_found += 1

        results.append({
            "video_path": csv_path,
            "frame_number": frame_number,
            "face_path": face_path,
            "confidence": float(confidence),
            "x1": x1,
            "y1": y1,
            "x2": x2,
            "y2": y2,
            "face_width": x2 - x1,
            "face_height": y2 - y1,
            "label": label,
            "label_name": label_name
        })

        frame_number += 1

    cap.release()

    print(
        f"{video_index + 1}/{len(test_df)} | "
        f"{label_name} | "
        f"frames checked: {frames_checked} | "
        f"faces saved: {faces_found}"
    )


# Save metadata
results_df = pd.DataFrame(results)

metadata_path = os.path.join(
    OUTPUT_DIR,
    "step5_test_metadata.csv"
)

results_df.to_csv(
    metadata_path,
    index=False
)

print("\n" + "=" * 60)
print("✅ BALANCED FACE EXTRACTION COMPLETE")
print("=" * 60)

print("Videos processed:", len(test_df))
print("Face crops saved:", len(results_df))
print("Metadata:", metadata_path)

print("\nFace crops by class:")
print(results_df["label_name"].value_counts())
```

```python
import cv2
import os
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

# Step 4 test settings
NUM_TEST_VIDEOS = 10
FRAME_INTERVAL = 10
MIN_CONFIDENCE = 0.90

metadata = []

test_df = df.head(NUM_TEST_VIDEOS).copy()

print("Testing videos:", len(test_df))

for idx, row in tqdm(
    test_df.iterrows(),
    total=len(test_df)
):
    
    video_path = get_kaggle_video_path(
        row["video_path"]
    )

    if video_path is None or not os.path.exists(video_path):
        print("Video not found:", video_path)
        continue

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("Could not open:", video_path)
        continue

    frame_number = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        if frame_number % FRAME_INTERVAL != 0:
            frame_number += 1
            continue

        # Convert BGR → RGB
        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        # Detect faces
        boxes, probs = mtcnn.detect(rgb)

        if boxes is not None:

            for face_idx, (box, prob) in enumerate(
                zip(boxes, probs)
            ):

                if prob is None or prob < MIN_CONFIDENCE:
                    continue

                x1, y1, x2, y2 = box.astype(int)

                # Keep coordinates inside image
                h, w = frame.shape[:2]

                x1 = max(0, x1)
                y1 = max(0, y1)
                x2 = min(w, x2)
                y2 = min(h, y2)

                if x2 <= x1 or y2 <= y1:
                    continue

                face = frame[
                    y1:y2,
                    x1:x2
                ]

                if face.size == 0:
                    continue

                face_width = x2 - x1
                face_height = y2 - y1

                # Save face
                face_filename = (
                    f"video_{idx}_"
                    f"frame_{frame_number}_"
                    f"face_{face_idx}.jpg"
                )

                face_path = os.path.join(
                    FACES_DIR,
                    face_filename
                )

                cv2.imwrite(
                    face_path,
                    face
                )

                metadata.append({
                    "video_path": row["video_path"],
                    "frame_number": frame_number,
                    "face_path": face_path,
                    "confidence": float(prob),
                    "x1": x1,
                    "y1": y1,
                    "x2": x2,
                    "y2": y2,
                    "face_width": face_width,
                    "face_height": face_height
                })

        frame_number += 1

    cap.release()


# Create metadata DataFrame
step4_df = pd.DataFrame(metadata)

# Save Step 4 metadata
STEP4_METADATA = os.path.join(
    STEP4_DIR,
    "step4_test_metadata.csv"
)

step4_df.to_csv(
    STEP4_METADATA,
    index=False
)

print("\n" + "=" * 50)
print("STEP 4 TEST COMPLETE")
print("=" * 50)

print("Videos tested:", len(test_df))
print("Face crops saved:", len(step4_df))
print("Metadata:", STEP4_METADATA)
```

```python
import cv2
import numpy as np
from tqdm.auto import tqdm

def calculate_blur_score(image_path):

    if not os.path.exists(image_path):
        return np.nan

    image = cv2.imread(image_path)

    if image is None:
        return np.nan

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    return float(
        cv2.Laplacian(
            gray,
            cv2.CV_64F
        ).var()
    )
```
